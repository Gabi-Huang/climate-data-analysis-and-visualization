```python
import pandas as pd
import seaborn as sns
import xarray as xr
from matplotlib import pyplot as plt

%matplotlib inline
```

```python
# load temperature data
temp = xr.open_dataset(r"C:\Users\USER\Desktop\MA\111-2\CAD\Week6\air.mon.mean.nc")
temp = temp.sel(time = slice('1970', '2020')) 
temp.load()
```

Taiwan rainfall

```python
temp_tai = temp.sel(lat = slice(26,20), lon = slice(118,125))
temp_tai_mean = temp_tai.mean(dim = ('lon', 'lat'))
temp_tai_mean
```

```python
month = temp_tai_mean['time.month']
year = temp_tai_mean['time.year']
yyyymm = year*100+month
yyyymm
```

```python
df1 = pd.DataFrame(yyyymm, columns=['YYYYMM'])
df2 = pd.DataFrame(year, columns=['YYYY'])
df3 = pd.DataFrame(month, columns=['MM'])
df4 = pd.DataFrame(temp_tai_mean['air'], columns=['air'])

df = pd.concat([df1, df2,df3,df4], axis=1, join='inner')
display(df)
```

```python
df['YYYYMM'] = pd.to_datetime(df1['YYYYMM'], format='%Y%m', errors='ignore')
df.set_index('YYYYMM', inplace=True)
```

```python
df1 = df.groupby(['YYYY', 'MM'],sort=False).agg(['mean'])
df1.columns = df1.columns.droplevel(0)
df1.reset_index(inplace = True)
df1.head()
```

```python
heatmap_data = pd.pivot_table(df1, values='mean', index=['MM'], columns='YYYY')
print(heatmap_data.iloc[0:3, 0:3])
```

```python
sns.heatmap(heatmap_data,cmap="coolwarm")
plt.xlabel("Year", size=14)
plt.ylabel("Month", size=14)
plt.title("Taiwan Temperature", size=14)
plt.tight_layout()
#plt.savefig('heatmap_with_Seaborn_python.jpg',dpi=150)
```
